## Tetraquarks & Pentaquarks

```python
from itertools import combinations_with_replacement
from groups.groups import Group, SO, SU, Sp
from groups.irreps import Irrep, get_rep, tensor_product, SumIrrep
```

```python
def hc_singlets(psi: Irrep, chi: Irrep, psibar=None, chibar=None, Nfermions: int = 4, HC: Group = SU(3)):
    if psibar:
        psis = (psi, psibar)
        str_psis = ('psi', 'psibar')
    else:
        psis = (psi, )
        str_psis = ('psi', )
    if chibar:
        chis = (chi, chibar)
        str_chis = ('chi', 'chibar')
    else:
        chis = (chi, )
        str_chis = ('chi', )

    combs = combinations_with_replacement((*psis, *chis), Nfermions)
    str_combs = combinations_with_replacement((*str_psis, *str_chis), Nfermions)

    for str_comb, comb in zip(str_combs, combs):
        try:
            all_irreps = tensor_product(*comb)
        except Exception as e:
            print(f'Could not calculate {str_comb}: {e}')
            continue

        boundstates = get_rep(all_irreps, HC.irrep(1))
        if boundstates:
            global_symmetries = SumIrrep(list(set(boundstates.summands)))
            print(*str_comb, ': ', global_symmetries)
        else:
            print(*str_comb, ':  no HC singlet')


def tetra_and_penta(psi, chi, HC, psibar=None, chibar=None):
    print('Tetraquarks:')
    hc_singlets(psi, chi, psibar=psibar, chibar=chibar, Nfermions=4, HC=HC)
    print()
    print('Pentaquarks:')
    hc_singlets(psi, chi, Nfermions=5, HC=HC)
```

## M1

```python
HC = SO(7)
psi = HC.irrep(7) * SU(5).irrep(5) * SU(6).irrep(1)
chi = HC.irrep(8) * SU(5).irrep(1) * SU(6).irrep(6)

tetra_and_penta(psi, chi, HC)
```
