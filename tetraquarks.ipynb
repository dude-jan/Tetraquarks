## Tetraquarks & Pentaquarks

```python
from itertools import combinations_with_replacement
from groups.groups import Simple, SO, SU, Sp, U1
from groups.irreps import Irrep, get_rep, tensor_product, SumIrrep
```

```python
def hc_singlets(psi: Irrep, chi: Irrep, Nfermions: int, symmetries: tuple[Simple, ...], psibar=None, chibar=None):
    HC, Gc, Hc, Gew, Hew = symmetries

    if psibar:
        psis = (psi, psibar)
        str_psis = ('psi', 'psibar')
    else:
        psis = (psi, )
        str_psis = ('psi', )
    if chibar:
        chis = (chi, chibar)
        str_chis = ('chi', 'chibar')
    else:
        chis = (chi, )
        str_chis = ('chi', )

    combs = combinations_with_replacement((*psis, *chis), Nfermions)
    str_combs = combinations_with_replacement((*str_psis, *str_chis), Nfermions)

    for str_comb, comb in zip(str_combs, combs):
        try:
            all_irreps = tensor_product(*comb)
        except Exception as e:
            print(f'Could not calculate {str_comb}: {e}')
            continue

        boundstates = get_rep(all_irreps, HC.irrep(1))
        if boundstates:
            global_symmetries = SumIrrep(list(set(boundstates.summands)))
            print(*str_comb, ': ', global_symmetries)
            if Hc != SU(3):
                su3reps = global_symmetries.remove(HC).decompose(Hc, decompose_from=Gc).decompose(SU(3)*U1(), decompose_from=Hc)
            else:
                su3reps = global_symmetries.remove(HC).decompose(Hc, decompose_from=Gc)
            su3reps = SumIrrep(list(set(su3reps.summands)))
            print(f'       -> {su3reps}')

            smreps = su3reps.decompose(Hew, decompose_from=Gew).decompose(SU(2, 'L') * SU(2, 'R'), decompose_from=Hew)
            print(f'       -> {SumIrrep(list(set(smreps.summands)))}')
            print()


def tetra_and_penta(psi, chi, Sgroups, psibar=None, chibar=None):
    print('Tetraquarks:')
    hc_singlets(psi, chi, 4, Sgroups, psibar=psibar, chibar=chibar)
    print()
    print('Pentaquarks:')
    hc_singlets(psi, chi, 5, Sgroups, psibar=psibar, chibar=chibar)
```

## M1

```python
HC = SO(7)
Gc = SU(6)
Gew = SU(5)
Sgroups = (HC, Gc, SO(6), Gew, SO(5))

psi = HC.irrep(7) * Gew.irrep(5) * Gc.irrep(1)
chi = HC.irrep(8) * Gew.irrep(1) * Gc.irrep(6)

tetra_and_penta(psi, chi, Sgroups)
```

## M5

```python
HC = Sp(4)
Gc = SU(6)
Gew = SU(5)
Sgroups = (HC, Gc, Sp(6), Gew, SO(5))

psi = HC.antisymmetric() * SU(5).irrep(5) * SU(6).irrep(1)
chi = HC.irrep(4) * SU(5).irrep(1) * SU(6).irrep(6)

tetra_and_penta(psi, chi, Sgroups)
```

## M6

```python
HC = SU(4)
Gc = SU(3, '0') * SU(3, '1')
Gew = SU(5)
Sgroups = (HC, Gc, SU(3), Gew, SO(5))

psi = HC.antisymmetric() * Gew.irrep(5) * Gc.irrep([1,1]) * U1().irrep(0)
chi = HC.irrep(4) * Gew.irrep(1) * Gc.irrep([3,1]) * U1().irrep(1)
chibar = HC.irrep(-4) * Gew.irrep(1) * Gc.irrep([1,-3]) * U1().irrep(-1)

tetra_and_penta(psi, chi, Sgroups, chibar=chibar)
```

# M12

```python
HC = SU(5)
Gc = SU(3, '0') * SU(3, '1')
Gew = SU(4, '0') * SU(4, '1')
Sgroups = (HC, Gc, SU(3), Gew, SU(4))

psi = HC.irrep(5) * SU(4, '0').irrep(4) * SU(4, '1').irrep(1) * SU(3, '0').irrep(1) * SU(3, '1').irrep(1) * U1('ew').irrep(1) * U1('x').irrep(0)
psibar = HC.irrep(-5) * SU(4, '0').irrep(1) * SU(4, '1').irrep(-4) * SU(3, '0').irrep(1) * SU(3, '1').irrep(1) * U1('ew').irrep(-1) * U1('x').irrep(0)
chi = HC.irrep(10) * SU(4, '0').irrep(1) * SU(4, '1').irrep(1) * SU(3, '0').irrep(3) * SU(3, '1').irrep(1) * U1('ew').irrep(0) * U1('x').irrep(1)
chibar = HC.irrep(-10) * SU(4, '0').irrep(1) * SU(4, '1').irrep(1) * SU(3, '0').irrep(1) * SU(3, '1').irrep(-3) * U1('ew').irrep(0) * U1('x').irrep(-1)

tetra_and_penta(psi, chi, Sgroups, psibar=psibar, chibar=chibar)
```
